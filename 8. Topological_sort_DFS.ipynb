#### Core idea: Fully explore each node's descendants first, then add the node itself to the order only after everything below it is done. Reverse the resulting "finish order" at the end.


```python
graph = {
    0: [1, 2],
    1: [3],
    2: [3],
    3: [4],
    4: []
}
```

```python
def topo_sort_dfs(graph):
    visited=set()
    topo=[]

    def dfs(u):
        visited.add(u)
        for v in graph[u]:
            if v not in visited:
                dfs(v)

        topo.append(u) #add AFTER fully exploring al neighbours

    for node in graph:
        if node not in visited:
            dfs(node)

    return topo[::-1] # reverse to get correct order

print(topo_sort_dfs(graph))

```
