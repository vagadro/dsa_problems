# Rotate Array by K Places

Given an integer array nums and a non-negative integer k, rotate the array to the left by k steps.

```python
def rotate_by_k(nums,k):
    n=len(nums)
    if n==0:
        return nums
    k=k%n
    temp=nums[:k]

    for i in range(k,n):
        nums[i-k]=nums[i]

    for i in range(k):
        nums[n-k+i ]=temp[i]
    return nums
```

```python
print(rotate_by_k([1, 2, 3, 4, 5, 6], k = 2))
print(rotate_by_k([3, 4, 1, 5, 3, -5], k = 8))
```
