# Length of loop in LL

Given the head of a singly linked list, find the length of the loop in the linked list if it exists. Return the length of the loop if it exists; otherwise, return 0.

A loop exists in a linked list if some node in the list can be reached again by continuously following the next pointer. Internally, pos is used to denote the index (0-based) of the node from where the loop starts.

Note that pos is not passed as a parameter.

```python
#Brute
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None


def loop_length_ll(head):
    visited={}
    curr=head
    position=0

    while curr is not None:
        if curr in visited:
            return position - visited[curr]
        visited[curr]=position
        position+=1
        curr=curr.next
    return 0


# Create nodes
n1 = ListNode(1)
n2 = ListNode(2)
n3 = ListNode(3)
n4 = ListNode(4)
n5 = ListNode(5)

# Connect nodes
n1.next = n2
n2.next = n3
n3.next = n4
n4.next = n5

# Create cycle: 5 points back to 3
n5.next = n3

head = n1

print(loop_length_ll(head))
```

```python
#Optimal
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None


def loop_length_optimal(head):
    slow=head
    fast=head

    # Phase 1: Detect Cucle
    while fast is not None and fast.next is not None:
        slow=slow.next
        fast=fast.next.next

        if slow==fast:
            break

    else:
        return 0

    # Phase 2: Count Loop length
    count=1
    curr=slow.next

    while curr!=slow:
        count+=1
        curr=curr.next
    
    return count


# Create nodes
n1 = ListNode(1)
n2 = ListNode(2)
n3 = ListNode(3)
n4 = ListNode(4)
n5 = ListNode(5)

# Connect nodes
n1.next = n2
n2.next = n3
n3.next = n4
n4.next = n5

# Create cycle: 5 points back to 3
n5.next = n3

head = n1

print(loop_length_optimal(head))
```
