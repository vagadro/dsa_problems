# Remove Nth Node From End of List

Given the head of a linked list, remove the nth node from the end of the list and return its head.

```python
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head=None

        
def delete_nth_node_from_end(head,n):

    if head is None:
        return None

    fast=head
    slow=head

    #move fast n steps ahead
    for i in range(n):
        fast=fast.next

    #if fast becomes None, delete the head- Case where n = length of LL
    if fast is None:
        return head.next
    

    #Now move both fast and slow one step at a time
    
    while fast.next is not None:
        slow=slow.next
        fast=fast.next

    slow.next=slow.next.next

    return head



# Traversal function
def traversal(head):
    curr = head

    while curr is not None:
        print(curr.val, end=" → ")
        curr = curr.next

    print("None")


n1 = ListNode(10)
n2 = ListNode(20)
n3 = ListNode(30)
n4 = ListNode(40)
n5 = ListNode(50)
n6 = ListNode(60)

n1.next = n2
n2.next = n3
n3.next = n4
n4.next = n5
n5.next = n6

head = n1

traversal(delete_nth_node_from_end(head,4))
```
