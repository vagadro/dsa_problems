# Doubly Linked List

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None

class DoublyLinkedList:
    def __init__(self):
        self.head=None

    def insert_at_start(self,data):
        new_node=Node(data)

        if self.head is None:
            self.head=new_node
            return 
        
        new_node.next=self.head
        self.head.prev=new_node

        self.head=new_node

    def insert_at_end(self,data):
        new_node=Node(50)

        if self.head is None:
            self.head=new_node
            return

        current=self.head

        while current.next is not None:
            current=current.next
        
        current.next=new_node
        new_node.prev=current

    def insert_at_position(self,data,position):
        if position <1:
            print("Invalid Position")
            return

        new_node=Node(data)

        # Case1: Insert at position 1
        if position==1:
            if self.head is None:
                self.head=new_node
                return
            
            new_node.next=self.head
            self.head.prev=new_node
            self.head=new_node
            return

        # Case 2: Not at the beginning and at somewhere mid
        current=self.head
        previous=None
        count=1

        # Reach the node currently at the requested position    
        while current is not None and count<position:
            previous=current
            current=current.next
            count+=1

        # If position is beyond end

        if current is None:
            if count==position:
                previous.next=new_node
                new_node.prev=previous

            else:
                print("Invalid Position")
            return

        # Insert before current
        new_node.next=current
        new_node.prev=previous

        previous.next=new_node
        current.prev=new_node


    def traverse(self):
        current=self.head

        while current:
            print(current.data, end=" ")
            current=current.next
        print("None")
```

```python
dll=DoublyLinkedList()
dll.traverse()
dll.insert_at_start(10)
dll.traverse()
dll.insert_at_end(50)
dll.traverse()
dll.insert_at_position(5,1)
dll.traverse()
dll.insert_at_position(20,3)
dll.traverse()
```
