# Find all the subsequences

```python
def generate_subsequences(nums):
    result=[]

    def solve(index,current):
        if index==len(nums):
            result.append(current.copy())
            return

        # Pick the element
        current.append(nums[index])
        solve(index+1,current)

        #Backtrack
        current.pop()

        #Do not pick the element
        solve(index+1,current)

    solve(0,[])
    return result


nums=[3,1,2]
print(generate_subsequences(nums))

```
