# Generate All Subsequences with K Sum

```python
def generate_subseq_k_sum(nums,k):
    current=[]

    def solve(index,current_sum):
        if index==len(nums):
            if current_sum==k:
                print(current.copy())
            return

        #Pick
        current.append(nums[index])
        solve(index+1,current_sum+nums[index])

        #Backtrack
        current.pop()

        #Not Pick
        solve(index+1,current_sum)

    solve(0,0)

generate_subseq_k_sum([1,2,1],2)


```
