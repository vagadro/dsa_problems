# Check if a Subset with Sum K exists

```python
def check_subset_sum(nums,k):
    def solve(index,current_sum):
        if index==len(nums):
            if current_sum==k:
                return True
            return False

        # Pick 
        if solve(index+1, current_sum+nums[index]):
            return True

        # Not pick
        if solve(index+1, current_sum):
            return True

        return False

    return solve(0,0)

print(check_subset_sum([1,2,3,4],6))
```
