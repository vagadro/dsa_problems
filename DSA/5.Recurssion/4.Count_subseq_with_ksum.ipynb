# Count all Subsequences with Sum K

```python
def count_subseq_sum_k(nums,k):
    current=[]

    def solve(index, current_sum):
        if index==len(nums):
            if current_sum==k:
                return 1
            return 0

        #Pick
        current.append(nums[index])
        pick=solve(index+1, current_sum+nums[index])

        #Bacjtrack
        current.pop()

        #Note pick
        not_pick=solve(index+1, current_sum)

        return pick + not_pick

    return solve(0,0)

print(count_subseq_sum_k([1,2,1],2))
```
