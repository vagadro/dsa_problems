# Generate All Binary Strings without Consecutive Ones

```python
def generate_binary_strings(n):
    current=[]
    result=[]

    def solve(index):
        if index==n:
            result.append("".join(current))
            return

        # Pick 0 (always allowed)
        current.append("0")
        solve(index+1)
        current.pop()

        #Pick 1 (only if previous character is not 1)
        #not current means if list is currently empty
        if not current or current[-1]!="1":
            current.append("1")
            solve(index+1)
            current.pop()

    solve(0)
    return result

print(generate_binary_strings(3))
```
