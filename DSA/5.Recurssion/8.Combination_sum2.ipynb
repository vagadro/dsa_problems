# Combination Sum2

Given a collection of candidate numbers (candidates) and a target number (target), find all unique combinations in candidates where the candidate numbers sum to target.

Each number in candidates may only be used once in the combination.

Note: The solution set must not contain duplicate combinations.



```python
def combination_sum2(candidates,target):
    # we first sort the candidates because sorting puts the same candidates next to each other. This makes it easier to skip when we choose not pick branch
    candidates.sort()
    
    current=[]
    result=[]

    def solve(index,current_sum):

        #Base Case1
        if current_sum==target:
            result.append(current.copy())
            return

        #Base case 2
        if index==len(candidates) or current_sum>target:
            return

        #Pick
        current.append(candidates[index])
        solve(index+1, current_sum+ candidates[index])

        #Backtrack
        current.pop()

        #Not pick- before not pick ensure that there are no duplicates
        while index+1<len(candidates) and candidates[index]==candidates[index+1]:
            index+=1

        solve(index+1,current_sum)
    solve(0,0)
    return result


```

```python
print(combination_sum2([10,1,2,7,6,1,5],8))
```

```python
print(combination_sum2([2,5,2,1,2],5))
```
