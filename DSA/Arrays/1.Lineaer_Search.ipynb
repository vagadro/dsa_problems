# Linear Search

Given an array arr of size n and a number target, return the index of the first occurrence of target. If it isn't present, return -1.

Example: arr = [4, 7, 2, 9, 5], target = 9 → Output: 3

```python
def linear_search(arr,target):
    n=len(arr)
    for i in range(len(arr)):
        if arr[i]==target:
            return i
    return -1

print(linear_search([4,7,2,9,5],9))
print(linear_search([4,7,2,9,5],6))
```
