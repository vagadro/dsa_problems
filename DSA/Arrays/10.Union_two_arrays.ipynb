# Union of two sorted arrays

Given two sorted arrays nums1 and nums2, return an array that contains the union of these two arrays. The elements in the union must be in ascending order.

The union of two arrays is an array where all values are distinct and are present in either the first array, the second array, or both.

```python
# Brute Force

def union_two_array_brute(nums1, nums2):
    union=set()

    for i in range(len(nums1)):
        union.add(nums1[i])

    for j in range(len(nums2)):
        union.add(nums2[j])

    return sorted(union)
```

```python
print(union_two_array_brute([1, 2, 3, 4, 5],[1, 2, 7]))
print(union_two_array_brute([3,4,6,7,9,9],[1,5,7,8,8]))
```

```python
def union_two_arrays_optimal(nums1, nums2):
    n=len(nums1)
    m=len(nums2)
    i=0
    j=0
    union=[]

    while i<n and j<m:
        if nums1[i]<=nums2[j]:
            if len(union)==0 or union[-1]!=nums1[i]:
                union.append(nums1[i])
            i+=1
        else:
            if len(union)==0 or union[-1]!=nums2[j]:
                union.append(nums2[j])
            j+=1

        
    # add leftover elements
    while i<n:
        if len(union)==0 or union[-1]!=nums1[i]:
            union.append(nums1[i])
        i+=1

    while j<m:
        if len(union)==0 or union[-1]!= nums2[j]:
            union.append(nums2[j])
        j+=1

    return union
```

```python
print(union_two_arrays_optimal([1, 2, 3, 4, 5],[1, 2, 7]))
print(union_two_arrays_optimal([3,4,6,7,9,9],[1,5,7,8,8]))
```
