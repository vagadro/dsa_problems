# Intersection of two sorted arrays

Given two sorted arrays, nums1 and nums2, return an array containing the intersection of these two arrays. Each element in the result must appear as many times as it appears in both arrays; that is, if an element appears x times in nums1 and y times in nums2, it should appear min(x, y) times in the result.

The intersection of two arrays is an array where all values are present in both arrays.

```python
# Brute Force
# for each element in array 1, check if equal element in array 2 exists. 


def inter_two_array_brute(nums1,nums2):
    n=len(nums1)
    m=len(nums2)
    visited=[0]*m
    result=[]

    for i in range(n):
        for j in range(m):
            if nums1[i]==nums2[j] and visited[j]==0:
                result.append(nums1[i])
                visited[j]=1
                break
    return result


print(inter_two_array_brute([1, 2, 2, 3, 5],[1, 2, 7]))
print(inter_two_array_brute([1, 2, 2, 3, 3, 3],[2, 3, 3, 4, 5, 7]))

```

```python
# Two pointer approach

# In this approach the point is since the arrays are sorted, for if nums1[i] is less than nums2[j] then we will increment i- this is because since the arrays are sorted, the value of i is less thatn values of j which means there is not intersection and they do not match. 

def inter_two_array_optimal(nums1,nums2):
    n=len(nums1)
    m=len(nums2)
    i=0
    j=0

    result=[]

    while i<n and j<m:
        if nums1[i] < nums2[j]:
            i+=1
        elif nums2[j] < nums1[i]:
            j+=1
        else:
            result.append(nums1[i])
            i+=1
            j+=1

    return result
    
```

```python
print(inter_two_array_optimal([1, 2, 2, 3, 5],[1, 2, 7]))
print(inter_two_array_optimal([1, 2, 2, 3, 3, 3],[2, 3, 3, 4, 5, 7]))
```
