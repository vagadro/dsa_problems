# Leaders in an Array

Given an integer array nums, return a list of all the leaders in the array.

A leader in an array is an element whose value is strictly greater than all elements to its right in the given array. The rightmost element is always a leader. The elements in the leader array must appear in the order they appear in the nums array.

```python
# Brute

def leaders_brute(nums):
    n=len(nums)
    result=[]

    for i in range(n):
        is_leader=1
        for j in range(i+1,n):
            if nums[j]>=nums[i]:
                is_leader=0
                break
        if is_leader==1:
            result.append(nums[i])

    return result

print(leaders_brute([1,2,5,3,1,2]))
print(leaders_brute([-3, 4, 5, 1, -4, -5]))
```

```python
# Optimal solutionm is to check the max from the right hand side

def leaders_optimal(nums):
    n=len(nums)
    result=[]
    max_right=float('-inf')

    for i in range(n-1,-1,-1):
        if nums[i] > max_right:
            result.append(nums[i])
            max_right=nums[i]

    result.reverse()
    return result

print(leaders_optimal([1,2,5,3,1,2]))
print(leaders_optimal([-3, 4, 5, 1, -4, -5]))
```
