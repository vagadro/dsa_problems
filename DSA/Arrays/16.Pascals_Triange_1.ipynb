# Pascal's Triangle I

Given two integers r and c, return the value at the rth row and cth column (1-indexed) in a Pascal's Triangle.

In Pascal's triangle:

The first row contains a single element 1.
Each row has one more element than the previous row.
Every row starts and ends with 1.

```python
def pascal_triangle_brute(r,c):
    triangle=[]

    for i in range(r):
        row=[1] * (i + 1)
        for j in range(1,i):
            row[j]=triangle[i-1][j-1] + triangle[i-1][j]
        triangle.append(row)
    return triangle[r-1][c-1]


print(pascal_triangle_brute(4,2))
```

Another approch is to figure out a pattern in Ncr


C(4, 2) = 4! / (2! × 2!) = (4 × 3 × 2 × 1) / ((2 × 1) × (2 × 1))
                         = (4 × 3) / (2 × 1)


- This means that the numerator keeps on decreasing by i
- Denominator keeps on increasing by 1

```python
def pascal_triangle_optimal(r,c):
    n=r-1
    k=c-1
    result=1

    for i in range(k):
        result=result * (n-i)
        result=result//(i+1)

    return result

print(pascal_triangle_optimal(4,2))
```
