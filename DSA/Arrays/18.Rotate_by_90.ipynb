# Rotate a Matrix 90 Degrees Clockwise In Place


Problem Statement

Given a square integer matrix "matrix" of size n x n. The matrix represents values arranged in rows and columns, and the required rotation is 90 degrees clockwise. Rotate matrix in-place.

Return nothing after the rotation.

```python
def rotate(matrix):
    n=len(matrix)
    ans = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            ans[j][n-1-i]=matrix[i][j]

    #copy back everyting to matrix 

    for i in range(n):
        for j in range(n):
            matrix[i][j]=ans[i][j]
    return matrix

print(rotate([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
```

```python
# Optimal Approach- first take the transpose of the matrix and reverse each row then. 

def rotate_optimal(matrix):
    n=len(matrix)

    # Step 1: Transpose
    for i in range(n):
        for j in range(i+1,n):
            matrix[i][j],matrix[j][i]=matrix[j][i],matrix[i][j]

    #step 2: Reverse eac row

    for i in range(n):
        matrix[i].reverse()

    return matrix

print(rotate_optimal([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
```
