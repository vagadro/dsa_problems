#  Set Matrix Zeroes

Given an m x n integer matrix matrix, if an element is 0, set its entire row and column to 0. You must do it in place.

```python
# Brute Approach

def set_matrix_zero(matrix):
    rows=len(matrix)
    cols=len(matrix[0])

    zero_rows=set()
    zero_cols=set()

    #find the rows and cols with 0 first
    for i in range(rows):
        for j in range(cols):
            if matrix[i][j]==0:
                zero_rows.add(i)
                zero_cols.add(j)

    #Now make the respectve rows and cols as 0

    for i in range(rows):
        for j in range(cols):
            if i in zero_rows or j in zero_cols:
                matrix[i][j]=0

    return matrix

print(set_matrix_zero([[1,1,1],[1,0,1],[1,1,1]]))
print(set_matrix_zero([[0,1,2,0],[3,4,5,2],[1,3,1,5]]))


# Time= O(n2)
#space=O(n+m) for two sets
```
