# Largest Element

Given an array of integers nums, return the value of the largest element in the array



Example: arr = [3, 2, 1, 5, 2] → Output: 5

```python
def largest_element_brute(arr):
    sort_arr=sorted(arr)
    return sort_arr[-1]


print(largest_element_brute([3,2,1,5,7]))
```

```python
def largest_element_optimal(arr):
    n=len(arr)
    largest=arr[0]

    for i in range(1,n):
        if arr[i]>=largest:
            largest=arr[i]

    return largest

print(largest_element_optimal([3,2,1,5,7])) 
```
