# 3 Sum

Given an integer array nums. Return all triplets such that:

i != j, i != k, and j != k
nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets. One element can be a part of multiple triplets. The output and the triplets can be returned in any order.


```python
def three_sum_brute(nums):
    n=len(nums)
    triplet_set=set()

    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                if nums[i] + nums[j] + nums[k]==0:
                    triplet=sorted([nums[i],nums[j],nums[k]])
                    triplet_set.add(tuple(triplet))
    
    final_result=[]
    for t in triplet_set:
        final_result.append(list(t))
    return final_result


print(three_sum_brute([2, -2, 0, 3, -3, 5])) 
```

```python
# Better lock two and serach remaining 1

def three_sum_better(nums):
    n=len(nums)
    triplet_set=set()

    for i in range(n):
        seen=set()
        for j in range(i+1,n):
            third=-(nums[i] + nums[j])
            if third in seen:
                triplet=sorted([nums[i],nums[j],third])
                triplet_set.add(tuple(triplet))
            else:
                seen.add(nums[j]) #add the last j seen

    final_result=[]
    for t in triplet_set:
        final_result.append(list(t))
    return final_result

print(three_sum_better([2, -2, 0, 3, -3, 5]))   

```

```python
# OPTIMAL--> SORT NUMS + TWO POINTERS

def three_sum_optimal(nums):
    n=len(nums)
    nums.sort()
    result=[]

    for i in range(n):
        #skip duplicate values for the first numers
        if i>0 and nums[i]==nums[i-1]:
            continue
        j=i+1
        k=n-1

        while j<k:
            total=nums[i] + nums[j] + nums[k]

            if total<0:
                j+=1
            elif total>0:
                k-=1
            else:
                result.append([nums[i], nums[j], nums[k]])
                j+=1
                k-=1

                #now if j and k are seen duplicate skip them
                while j<k and nums[j]==nums[j-1]:
                    j+=1
                while j<k and nums[k]==nums[k+1]:
                    k-=1
    return result

print(three_sum_optimal([2, -2, 0, 3, -3, 5]))  
```
