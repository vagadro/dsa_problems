# FOur Sum

Given nums and a target, return all unique quadruplets [a, b, c, d] (from four different indices) where a + b + c + d = target. No duplicate quadruplets in the answer; order doesn’t matter.

Example:

nums = [1, 0, -1, 0, -2, 2], target = 0
Output: [[-2, -1, 1, 2], [-2, 0, 0, 2], [-1, 0, 0, 1]]

```python
def four_sum_brute(nums,target):
    n=len(nums)
    result_set=set()

    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                for l in range(k+1,n):
                    if nums[i] + nums[j]+ nums[k]+nums[l]==0:
                        quad=sorted([nums[i],nums[j],nums[k],nums[l]])
                        result_set.add(tuple(quad))

    final_result=[]
    for t in result_set:
        final_result.append(list(t))
    return final_result


print(four_sum_brute([1, 0, -1, 0, -2, 2],0))
```

```python
# Better approach- Fix two and hasing for the rest

def four_sum_better(nums,target):
    n=len(nums)
    result_set=set()

    for i in range(n):
        for j in range(i+1,n):
            seen=set()
            for k in range(j+1,n):
                fourth=target- (nums[i]+nums[j]+nums[k])
                if fourth in seen:
                    quad=sorted([nums[i],nums[j],nums[k],fourth])
                    result_set.add(tuple(quad))
                else:
                    seen.add(nums[k])

    final_result=[]
    for t in result_set:
        final_result.append(list(t))
    return final_result


print(four_sum_better([1, 0, -1, 0, -2, 2],0))
```

```python
# Optimal - Just like 3 sum skip the duplicates

def four_sum_optimal(nums,target):
    n=len(nums)
    nums.sort()
    result=[]

    for i in range(n):
        #skip duplicates first numners
        if i>0 and nums[i]==nums[i-1]:
            continue

        for j in range(i+1,n):
            #skip diplicate second numbers
            if j>i+1 and nums[j]==nums[j-1]:
                continue

            k=j+1
            l=n-1

            while k<l:
                total=nums[i] + nums[j] + nums[k] + nums[l]

                if total<target:
                    k+=1
                elif total>target:
                    l-=1
                else:
                    result.append([nums[i],nums[j],nums[k],nums[l]])
                    k+=1
                    l-=1
                    while k<1 and nums[k]==nums[k-1]:
                        k+=1
                    while k<1 and nums[l]==nums[l+1]:
                        l-=1
    return result

print(four_sum_better([1, 0, -1, 0, -2, 2],0))
```
