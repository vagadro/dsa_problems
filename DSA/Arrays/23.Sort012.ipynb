# Sort 0s, 1s and 2s- Dutch National Flag

Problem (LeetCode 75, “Sort Colors”): nums contains only 0s, 1s and 2s. Sort it in place so all 0s come first, then 1s, then 2s. (LeetCode asks you not to use the built-in sort.)

```python
def sort012_brute(nums):
    nums.sort()
    return nums

print(sort012_brute([2, 0, 2, 1, 1, 0]))
```

```python
# Better

def sort012_better(nums):
    n=len(nums)
    count0=0
    count1=0
    count2=0

    result=[]

    for i in range(n):
        if nums[i]==0:
            count0+=1
        elif nums[i]==1:
            count1+=1
        else:
            count2+=1

    for i in range(count0):
        result.append(0)
    for i in range(count1):
        result.append(1)
    for i in range(count2):
        result.append(2)

    return result

print(sort012_better([2, 0, 2, 1, 1, 0]))
```

```python
# Optimal Dutch National Flag
# have three pointsers= left to check 0, mid to check 1 and right to check 2

def sort012_optimal(nums):
    low=0
    mid=0
    right=len(nums)-1

    while mid<=right:
        if nums[mid]==0:
            nums[low],nums[mid]=nums[mid],nums[low]
            low+=1
            mid+=1
        elif nums[mid]==1:
            mid+=1
        else:
            nums[mid],nums[right] = nums[right],nums[mid]
            right -=1
    return nums

print(sort012_optimal([2, 0, 2, 1, 1,0]))
```
