# Kadane’s Algorithm: Maximum Subarray Sum

Problem (LeetCode 53): Given nums, find the contiguous subarray (at least one element) with the largest sum, and return that sum.

A subarray is a continuous slice of the array. [4, -1, 2] is a subarray of [1, 4, -1, 2, 5], but [4, 2] is not (it skips -1).

```python
def subarray_sum_brute(nums):
    n=len(nums)
    max_sum=float('-inf')

    for i in range(n):
        total=0
        for j in range(i,n):
            total+=nums[j]
            if total>max_sum:
                max_sum=total
    return max_sum


print(subarray_sum_brute([2, 3, 5, -2, 7, -4]))
print(subarray_sum_brute([-2, -3, -7, -2, -10, -4]))
```

```python
# Optimal approch
# Key insight: Walk through the array once, keeping a running sum current. If current ever becomes negative, it can only hurt whatever comes next (adding a negative number makes any future sum smaller). So throw it away and start fresh from the next element by resetting current = 0.


def subarray_sum_optimal(nums):
    n=len(nums)
    max_sum=float('-inf')
    current=0

    for i in range(n):
        current += nums[i]

        if current > max_sum:
            max_sum=current

        if current<0:
            current=0
    return max_sum

print(subarray_sum_optimal([2, 3, 5, -2, 7, -4]))
print(subarray_sum_optimal([-2, -3, -7, -2, -10, -4]))
```
