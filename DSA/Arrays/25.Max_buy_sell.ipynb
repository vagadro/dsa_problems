# Best Time to Buy and Sell Stock

Problem (LeetCode 121): prices[i] is the stock price on day i. Choose one day to buy and a later day to sell. Return the maximum profit. If no profit is possible, return 0.

```python
def buy_sell_brute(prices):
    n=len(prices)
    max_profit=0

    for i in range(n):
        for j in range(i+1,n):
            profit=prices[j]-prices[i]
            max_profit=max(max_profit, profit)

    return max_profit

print(buy_sell_brute([7,1,5,3,6,4]))

```

```python
def buy_sell_optimal(prices):
    min_price=float('inf')
    max_profit=0
    n=len(prices)

    for i in range(n):
        min_price=min(min_price, prices[i])
        profit=prices[i] - min_price
        max_profit=max(max_profit, profit)
    return max_profit

print(buy_sell_optimal([7,1,5,3,6,4]))
```
