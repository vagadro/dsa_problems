# Longest Consecutive Sequence in an Array

Given an array nums of n integers.

Return the length of the longest sequence of consecutive integers. The integers in this sequence can appear in any order.



```python
def longest_cons_sequence_brute(nums):
    n=len(nums)
    longest=0

    for num in nums:
        current=num
        count=1

        while current+1 in nums:
            current+=1
            count+=1
        longest=max(longest, count)

    return longest
print(longest_cons_sequence_brute([100, 4, 200, 1, 3, 2]))
print(longest_cons_sequence_brute([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))
```

```python
# Better approach is to sort and count

def longest_cons_sequence_better(nums):
    if len(nums)==0:
        return 0

    nums.sort()
    last_smaller=float('-inf')
    count=0
    longest=0

    for num in nums:
        if num - 1 == last_smaller:
            count+=1
            last_smaller=num
        elif num != last_smaller:
            count=1
            last_smaller=num
        longest= max(longest,count)

    return longest

print(longest_cons_sequence_better([100, 4, 200, 1, 3, 2]))
print(longest_cons_sequence_better([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))
```

```python
# Here we will use set

def longest_cons_sequence_optimal(nums):
    if len(nums)==0:
        return 0

    num_set=set(nums)
    longest=0

    for num in num_set:
        if num -1 not in num_set:
            x=num
            count=1

            while x + 1 in num_set:
                x+=1
                count+=1
            longest=max(longest,count)
    return longest

print(longest_cons_sequence_optimal([100, 4, 200, 1, 3, 2]))
print(longest_cons_sequence_optimal([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))
```
