# Longest Subarray with Sum K

Problem: Given nums and an integer k, return the length of the longest subarray (contiguous) whose elements add up to exactly k. Return 0 if there is none.

```python
def longest_subarray_sum_brute(nums,k):
    n=len(nums)
    longest=0

    for i in range(n):
        total=0
        for j in range(i,n):
            total+=nums[j]

            if total==k:
                longest=max(longest, j-i +1)
    return longest

print(longest_subarray_sum_brute([10, 5, 2, 7, 1, 9],15))
print(longest_subarray_sum_brute([-3,2,1],6))
```

```python
# Prefix Sum + Hashings

def longest_subarray_sum_optimal(nums, k):
    prefix_sum=0
    prefix_map={}  #index where prefix_sum appreared
    longest=0

    for i in range(len(nums)):
        prefix_sum+=nums[i]

        #Case 1: the subarray from 0 to i iteself sums to k
        if prefix_sum==k:
            longest=max(longest, i+1)
        
        # CAse 2: A previous prefix sum lets us find a subarray of sum k
        required=prefix_sum-k

        if required in prefix_map:
            length= i - prefix_map[required]
            longest=max(longest, length)

        if prefix_sum not in prefix_map:
            prefix_map[prefix_sum]=i

    return longest

print(longest_subarray_sum_optimal([10, 5, 2, 7, 1, 9],15))
print(longest_subarray_sum_optimal([-3,2,1],6))
```
