# Largest Subarray with Sum 0

Problem: Given nums (which can contain positive numbers, negative numbers and zeros), return the length of the longest subarray whose elements add up to 0. Return 0 if there is none.

```python
def largest_subarray_sum0_brute(nums):
    max_length=0

    for i in range(len(nums)):
        curr_sum=0
        for j in range(i, len(nums)):
            curr_sum+= nums[j]
            if curr_sum==0:
                max_length=max(max_length, j-i+1)
    return max_length

print(largest_subarray_sum0_brute([15, -2, 2, -8, 1, 7, 10, 23]))
print(largest_subarray_sum0_brute([2,10,4]))
```

```python
def largest_subarray_sum0_optimal(nums):
    prefix_sum=0
    prefix_map={}
    max_length=0


    for i in range(len(nums)):
        prefix_sum+=nums[i]

        #Case 1: the subaary from 0 to i sums to 0
        if prefix_sum==0:
            max_length=max(max_length, i+1)

        #CAse 2: the prefix_sum appread before in prefix map

        if prefix_sum in prefix_map:
            length= i - prefix_map[prefix_sum]
            max_length= max(max_length, length)

        if prefix_sum not in prefix_map:
            prefix_map[prefix_sum]= i
    return max_length

print(largest_subarray_sum0_optimal([15, -2, 2, -8, 1, 7, 10, 23]))
print(largest_subarray_sum0_optimal([2,10,4]))
```
