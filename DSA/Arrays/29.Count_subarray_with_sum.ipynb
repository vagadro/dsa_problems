# Count Subarrays with Sum K

Problem (LeetCode 560): Given nums (which can contain negatives and zeros) and an integer k, return the number of subarrays whose elements add up to exactly k.

```python
def count_subarray_sum(nums,k):
    count=0

    for i in range(len(nums)):
        curr_sum=0
        for j in range(i, len(nums)):
            curr_sum+= nums[j]
            if curr_sum == k:
                count+=1
    return count

print(count_subarray_sum([1,1,1],2))
print(count_subarray_sum([1,2,3],3))
```

```python
def count_subarray_sum_optimal(nums,k):
    prefix_sum=0
    prefix_map={0:1}
    count=0


    for i in range(len(nums)):
        prefix_sum+=nums[i]

        required= prefix_sum-k

        if required in prefix_map:
            count+= prefix_map[required]
        
        #Record the current prefiix sum (after checking)
        if prefix_sum in prefix_map:
            prefix_map[prefix_sum]+=1
        else:
            prefix_map[prefix_sum]=1

    return count

print(count_subarray_sum_optimal([1,1,1],2))
print(count_subarray_sum_optimal([1,2,3],3))
```
