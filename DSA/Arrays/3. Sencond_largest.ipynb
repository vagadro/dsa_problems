# Second Largest Element

Given an array arr of size n, return the second largest distinct element. If it doesn't exist (for example, all elements are equal), return -1.

Example: arr = [1, 2, 4, 7, 7, 5] → Output: 5

```python
def second_largest_brute(nums):
    sort_nums=sorted(nums)
    largest=sort_nums[-1]
    n=len(nums)

    for i in range(n-1,0,-1):
        if nums[i]<largest:
            return i
    return -1

print(second_largest_brute([1,2,4,7,7,5]))

```

Time- O(nlogn) sort
space~0(n)

# BEtter Approach -2 Passes

```python
def second_largest_better(nums):
    n=len(nums)
    if n<2:
        return -1

    #Pass 1: Largest
    largest=nums[0]
    for i in range(n):
        if nums[i]>largest:
            largest=nums[i]
        
    # PAss 2: find 2nd largest now
    second=float('-inf')
    for i in range(n):
        if nums[i]>second and nums[i]<largest:
            second=nums[i]
    
    if second==float('-inf'):
        return -1
    return second

```

```python
print(second_largest_better([1,2,4,7,7,5]))
```

# Optimal - 2 pointer and single pass

```python
def second_largest_optimal(nums):
    n=len(nums)
    if n<2:
        return -1

    largest=float('-inf')
    second=float('-inf')

    for i in range(n):
        if nums[i]>largest:
            second=largest
            largest=nums[i]
        elif nums[i]<largest and nums[i]>second:
            second=nums[i]

    if second==float('-inf'):
        return -1
    return second

print(second_largest_optimal([1,2,4,7,7,5]))
```
