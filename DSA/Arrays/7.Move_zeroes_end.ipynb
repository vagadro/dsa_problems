# Move Zeroes to the end of Array

Given an integer array nums, move all the 0's to the end of the array. The relative order of the other elements must remain the same.

```python
def move_zeroes_end1(nums):
    insert_pos=0
    n=len(nums)


    #1st pass- move all non zeroes at the beginining using inser_pos
    for i in range(n):
        if nums[i]!=0:
            nums[insert_pos]=nums[i]
            insert_pos+=1
    
    #fill the rest of the positions with 0
    while insert_pos < n:
        nums[insert_pos]=0
        insert_pos+=1
    return nums
```

```python
print(move_zeroes_end1([0,1,4,0,5,2]))
```

```python
# Other approach is to use a i, j pointer- basically two pointer approach
# j pointer checks the 0
# i pointer increments the positions


def move_zeroes_end2(nums):
    n=len(nums)
    j=0 

    for i in range(n):
        if nums[i]!=0:
            nums[j],nums[i]=nums[i],nums[j]
            j+=1
    return nums
```

```python
print(move_zeroes_end2([0,1,0,3,12]))
print(move_zeroes_end2([0,0,1]))   
```
