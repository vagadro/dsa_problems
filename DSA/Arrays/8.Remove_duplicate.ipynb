# Remove Duplicates from An Array

```python
def remove_duplicate(nums):
    unique=sorted(set(nums))
    n=len(unique)

    for i in range(n):
        nums[i]=unique[i]
    return n
```

```python
print(remove_duplicate([0, 0, 3, 3, 5, 6]))
```

```python
# Another apporiach is Two pointers appraoch

# Put i as start index
# j as the index we will chec
# if nums[i]!=nums[j]--> this means a match, but you do not swap i and j, you swap i+1 and j. 
# this means you first increamtnet i and then swap

def remove_duplicate2(nums):
    n=len(nums)
    if n==0:
        return 0

    i=0
    for j in range(1,n):
        if nums[i] != nums[j]:
            i+=1
            nums[i] = nums[j]
    return i+1

print(remove_duplicate2([0, 0, 3, 3, 5, 6]))
```
