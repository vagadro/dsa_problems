# Binary Search

```python
def binary_search(nums, target):
    n=len(nums)
    low=0
    high=n-1

    while low<=high:
        mid= (low+ high)//2

        if nums[mid]==target:
            return mid
        elif target > mid:
            low=mid +1
        else:
            high=mid-1
    return None

print(binary_search([-1,0,3,5,9,12],9))
print(binary_search([-1,0,3,5,9,12],3))
```
