# Lower Bound

Problem: Given a sorted array nums and a target, return the lower bound: the smallest index i such that nums[i] >= target. If no element is >= target, return n (the length of the array, meaning “past the end”).

```python
def lower_bound_brute(nums,target):
    for i in range(len(nums)):
        if nums[i]>=target:
            return i
    return None

print(lower_bound_brute([1,2,2,3],2))
print(lower_bound_brute([3,5,8,15,19],9))
```

```python
def lower_bound_optimal(nums,target):
    low=0
    high=len(nums)
    answer=len(nums) #lets consider the larget numbers is the lower bound

    while low<=high:
        mid=(low+high)//2

        if nums[mid]>=target:
            answer=mid
            high=mid-1
        else:
            low=mid+1
    return answer

print(lower_bound_optimal([1,2,2,3],2))
print(lower_bound_optimal([3,5,8,15,19],9))
```
