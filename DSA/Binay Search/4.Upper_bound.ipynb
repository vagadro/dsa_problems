# Upper Bound

Problem: Given a sorted array nums and a target, return the upper bound: the smallest index i such that nums[i] > target (strictly greater). If no element is greater, return n.

```python
def upper_bound_brute(nums, target):
    for i in range(len(nums)):
        if nums[i]>target:
            return i
    return len(nums)

print(upper_bound_brute([1,2,2,3],2))
print(upper_bound_brute([3,5,8,15,19],9))
```

```python
def upper_bound_optimal(nums,target):
    low=0
    high=len(nums)-1
    answer=len(nums)

    while low<=high:
        mid= (low+ high)//2

        if nums[mid] > target:
            answer=mid
            high=mid -1
        else:
            low= mid +1
    return answer

print(upper_bound_optimal([1,2,2,3],2))
print(upper_bound_optimal([3,5,8,15,19],9))
```
