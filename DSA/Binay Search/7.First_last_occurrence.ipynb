# First and last occurrence

Given an array of integers nums sorted in non-decreasing order, find the starting and ending position of a given target value. If the target is not found in the array, return [-1, -1].

```python
def first_last(nums,target):
    n=len(nums)
    first_index=-1
    last_index=-1

    for i in range(n):
        if nums[i]==target:
            if first_index== -1:
                first_index=i
            last_index=i
    return [first_index, last_index]


print(first_last([5,7,7,8,8,10],8))
print(first_last([5,7,7,8,8,10],6))
print(first_last([],0))
```

```python
def lower_bound(nums,target):
    low=0
    high=len(nums)-1
    lower_index=len(nums)

    while low<=high:
        mid= (low+high)//2

        if nums[mid]>=target:
            lower_index=mid #found but still an index lower can exist
            high = mid-1
        else:
            low=mid+1
    return lower_index

def upper_bound(nums,target):
    low=0
    high=len(nums)-1
    upper_index=len(nums)

    while low<=high:
        mid = (low + high)//2

        if nums[mid] > target:
            upper_index= mid
            high=mid -1
        else:
            low = mid +1
    return upper_index

def first_last_optimal(nums,target):
    lower_index= lower_bound(nums,target)
    if lower_index== len(nums) or nums[lower_index]!=target:
        return [-1,-1]
    upper_index=upper_bound(nums,target)
    return [lower_index, upper_index-1]

print(first_last_optimal([5,7,7,8,8,10],8))
print(first_last_optimal([5,7,7,8,8,10],6))
print(first_last_optimal([],0))  
```
