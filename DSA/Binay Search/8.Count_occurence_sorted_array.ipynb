# Count Occurrences in a Sorted Array

Problem: Given a sorted array nums and a target, return how many times target appears. Return 0 if it doesn’t appear.

```python
def count_occurence_in_array(nums,target):
    n=len(nums)
    count=0
    for i in range(n):
        if nums[i]==target:
            count+=1
    return count

print(count_occurence_in_array([0, 0, 1, 1, 1, 2, 3],1))
print(count_occurence_in_array([0, 0, 1, 1, 1, 2, 3],4))
```

```python
def lower_bound(nums,target):
    low=0
    high=len(nums)-1
    lower_index=len(nums)

    while low<=high:
        mid=(low+high)//2

        if nums[mid]>=target:
            lower_index=mid
            high=mid-1
        else:
            low=mid + 1
    return lower_index

def upper_bound(nums,target):
    low=0
    high=len(nums)-1
    upper_index=len(nums)

    while low<=high:
        mid= (low+high)//2

        if nums[mid]>target:
            upper_index=mid
            high=mid-1

        else:
            low=mid+1
    return upper_index

def count_occurence(nums,target):
    lower_index=lower_bound(nums,target)

    if lower_index==-1:
        return 0
    upper_index=upper_bound(nums,target)
    return upper_index - lower_index + 1

```

```python
print(count_occurence([0, 0, 1, 1, 1, 2, 3],1))
print(count_occurence([0, 0, 1, 1, 1, 2, 3],4))
```
