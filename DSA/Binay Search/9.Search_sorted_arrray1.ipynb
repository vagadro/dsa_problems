# Search in rotated sorted array-I

Given an integer array nums, sorted in ascending order (with distinct values) and a target value k. The array is rotated at some pivot point that is unknown. Find the index at which k is present and if k is not present return -1.

```python
def search_brute(nums,target):
    for i in range(len(nums)):
        if nums[i]==target:
            return i
    return -1

print(search_brute([4, 5, 6, 7, 0, 1, 2], 0))
```

```python
# Optimal- modified Binary search- Search the mid and check if the left or right or mid array is sorted or not. The answer will lie in the sorted array

def search_in_sorted(nums,target):
    left=0
    right=len(nums) -1

    while left<=right:
        mid=(left + right)//2

        if nums[mid]==target:
            return mid

        # Check if the Left half is sorted

        if nums[left]<=nums[mid]:

            # Is target inside the sorted left half?
            if nums[left] <=target<nums[mid]:
                right=mid-1
            else:
                left=mid + 1

        # Otherise the right half must be sorted
        else:

            # Is the target inside the right sorted half?

            if nums[mid]<target <= nums[right]:
                left=mid+1
            else:
                right=mid-1

    return -1


print(search_in_sorted([4, 5, 6, 7, 0, 1, 2], 0))
```
