```python

class Node:
    def __init__(self,val):
        self.val=val
        self.next=None

class SinglyLinkedList:
    def __init__(self):
        self.head=None


    def append(self,val):
        new_node=Node(val)

        # Case 1: If list is currently empty
        if self.head==None:
            self.head=new_node

        # Case 2: If list is not empty
        else:
            curr=self.head
            while curr.next is not None:
                curr=curr.next

            # Once the pointer reach to last node update the new node to the list
            curr.next=new_node

    def traversal(self):

        #Case 1: If LL is empty
        if self.head is None:
            print("SLL is Empty")
            
        # Case 2: If list is not empty
        else:
            curr=self.head
            while curr is not None:
                print(curr.val, end=" ")
                curr=curr.next
            print()

sll=SinglyLinkedList()
sll.append(10)
sll.append(20)
sll.append(30)
sll.append(40)
sll.traversal()
```
