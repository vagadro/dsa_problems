# Middle of the Linked List

Given the head of a singly linked list, return the middle node of the linked list.

If there are two middle nodes, return the second middle node.



```python
class Node:
    def __init__ (self,val):
        self.val=val
        self.next=None


class SinglyLinkedList:
    def __init__ (self):
        self.head=None

    def append(self,val):
        new_node=Node(val)

        #Case 1: if list is currently empty

        if self.head is None:
            self.head=new_node
            return

        # Case 2: If list is not empty
        else:
            curr=self.head
            while curr.next is not None:
                curr=curr.next
            
            # Once the pointer reach to last node update the new node to the list
            curr.next=new_node


    def insert_at(self,val,pos):

        new_node=Node(val)

        #Case 1: insert at beginning
        if pos==0:
            new_node.next=self.head
            self.head=new_node

        else:
            curr=self.head
            prev_node=None
            count=0
            while curr is not None and count<pos:
                prev_node=curr
                curr=curr.next
                count+=1
            prev_node.next=new_node
            new_node.next=curr        


    def delete(self,val):
        temp=self.head

        # Case 1: empty LL
        if self.head is None:
            print("SLL is empty")
            return 

        #Case 2: Value exists in the head node

        if self.head.val==val:
            self.head=self.head.next
            return

        # Case 3: Search for the vaalue and then delete
        curr=self.head

        while curr.next is not None:

            if curr.next.val==val:
                curr.next=curr.next.next
                return
            curr=curr.next
        print("Value not found")

    def search(self,val):
        curr=self.head
        pos=0

        while curr is not None:
            if curr.val==val:
                return pos
            curr=curr.next
            pos+=1
        return -1

    
    def traversal(self):

        #Case 1: if the list is empty
        if self.head is None:
            print("SLL is empty")

        #Case 2: If the list is not empty print

        else:
            curr=self.head
            while curr is not None:
                print(curr.val, end=" ")
                curr=curr.next
            print()

    def count_nodes(self):

        count=0
        curr=self.head

        while curr is not None:
            count+=1
            curr=curr.next
        return count

    def find_middle(self):

        # first traversal: count nodes

        count=0
        curr=self.head

        while curr is not None:
            count+=1
            curr=curr.next

        #find middle position
        middle=count//2

        # Now once you found the middle, traverse and get the value
        curr=self.head

        for i in range(middle):
            curr=curr.next
        return curr

    def find_middle_slow_fast(self):
        slow=self.head
        fast=self.head

        while fast is not None and fast.next is not None:
            slow=slow.next
            fast=fast.next.next

        return slow
    
sll=SinglyLinkedList()
sll.append(10)
sll.append(20)
sll.append(30)
sll.append(40)
sll.insert_at(25,2)
sll.traversal()
```
