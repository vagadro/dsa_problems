# Reverse Linked List

Given the head of a singly linked list, reverse the list, and return the reversed list.

```python
class Node:
    def __init__(self,val):
        self.val=val
        self.next=None


class SinglyLinkedList:
    def __init__(self):
        self.head=None


    def append(self,val):
        new_node=Node(val)

        #Case 1: if list is currently empty

        if self.head is None:
            self.head=new_node
            return

        # Case 2: If list is not empty
        else:
            curr=self.head
            while curr.next is not None:
                curr=curr.next
            
            # Once the pointer reach to last node update the new node to the list
            curr.next=new_node



    def insert_at(self,val,pos):
        new_node=Node(val)

        #Case 1: insert at begining

        if pos==0:
            new_node.next=self.head
            self.head=new_node
        
        # Case 2: Not the the beginning of :: 
        else:
            curr=self.head
            prev_node=None
            count=0

            while curr is not None and count<pos:
                prev_node=curr
                curr=curr.next
                count+=1
            prev_node.next=new_node
            new_node.next=curr

    def delete(self,val):
        temp=self.head

        # Case 1: Empty LL
        if self.head is None:
            print("SLL is empty")
            return

        # Case 2: VAlue exsts at head node    
        if self.head.val==val:
            self.head=self.head.next
            return

        # Case 3: Saerch for value and then delete

        curr=self.head

        while curr.next is not None:
            if curr.next.val==val:
                curr.next=curr.next.next
                return
            curr=curr.next
        print("Value not found")


    def search(self, val):
        curr=self.head
        pos=0

        while curr is not None:
            if curr.val==val:
                return pos
            curr=curr.next
            pos+=1
        return -1

    def traversal(self):

        # Cas 1: if the list is emptty

        if self.head is None:
            print("SLL is empty")

        # Case 2: if the list is not empty print
        else:
            curr=self.head
            while curr is not None:
                print(curr.val, end=" ")
                curr=curr.next
            print()

    def reverse(self):
        prev=None
        curr=self.head

        while curr is not None:
            next_node=curr.next
            curr.next=prev

            prev=curr
            curr=next_node

        self.head=prev    
        
        return prev

sll=SinglyLinkedList()
sll.append(10)
sll.append(20)
sll.append(30)
sll.append(40)
sll.insert_at(25,2)
sll.traversal()
sll.reverse()          
sll.traversal()
```
