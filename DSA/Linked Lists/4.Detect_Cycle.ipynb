# Linked List Cycle

Given head, the head of a linked list, determine if the linked list has a cycle in it.

There is a cycle in a linked list if there is some node in the list that can be reached again by continuously following the next pointer. Internally, pos is used to denote the index of the node that tail's next pointer is connected to. Note that pos is not passed as a parameter.

Return true if there is a cycle in the linked list. Otherwise, return false.



```python
def hasCycleBrute(self,head):
    visited=set()
    curr=head

    while curr is not None:
        if curr in visited:
            return True
        visited.add(curr)
        curr=curr.next

    return False
```

```python
def hasCycleOptimal(self,head):
    slow=head
    fast=head

    while fast is not None and fast.next is not None:
        slow=slow.next
        fast=fast.next.next

        if slow==fast:
            return True
    return False
```

```python
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None


def has_cycle(head):

    visited = set()
    curr = head

    while curr is not None:

        if curr in visited:
            return True

        visited.add(curr)
        curr = curr.next

    return False


# Create nodes
n1 = ListNode(1)
n2 = ListNode(2)
n3 = ListNode(3)
n4 = ListNode(4)
n5 = ListNode(5)

# Connect nodes
n1.next = n2
n2.next = n3
n3.next = n4
n4.next = n5

# Create cycle: 5 points back to 3
n5.next = n3

head = n1
```

```python
result = has_cycle(head)
print(result)
```
