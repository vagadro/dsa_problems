# Course Scheduling2 is an extension of CS1 with Topological ORder (DFS)

- The only change here is instead of printing True/False yoiu return the Actual ORder possible

```python
def find_order(numCourses,prerequisites):
    graph={i:[] for i in range(numCourses)}
    for a,b in prerequisites:
        graph[a].append(b)

    visited=set()
    rec_stack=set()
    topo=[]

    def dfs(node):
        visited.add(node)
        rec_stack.add(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True
            elif neighbour in rec_stack:
                return True

        rec_stack.remove(node)
        topo.append(node)
        return False

    for node in range(numCourses):
        if node not in visited:
            if dfs(node):
                return []

    return topo[::-1]

prerequisites = [[1, 0], [0, 1]]
print(find_order(2,prerequisites))

prerequisites_ok = [[1, 0], [2, 1], [3, 2]]
print(find_order(4, prerequisites_ok))  
```
