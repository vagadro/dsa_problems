# Course Scheduling 3 BSF- Kahn's Algorithm

```python
1. Count how many incoming edges (in-degree) each node has
       
        ↓
2. Find all nodes with in-degree 0 (no prerequisites) → put them in a queue
        ↓
3. WHILE the queue is not empty:
       
        ↓
   a. Take a node out of the queue (it's ready — all its dependencies are met)
       
        ↓
   b. Add it to topo_order (this node's position in the final answer is locked in)
       
        ↓
   c. For each neighbour of this node:
        - Decrease that neighbour's in-degree by 1 
          (because we just "removed" this node's edge pointing to it)
        - If that neighbour's in-degree just hit 0,
          it now has no remaining prerequisites → add it to the queue
       
        ↓
4. Once the queue is empty:
   - If topo_order contains ALL nodes → valid topological order found ✅
   - If topo_order is missing some nodes → there's a cycle, no valid order exists 
```

```python
from collections import deque


def course_schedule_kahn_algo(num_nodes,course):
    graph={i:[] for i in range(num_nodes)}
    in_degree={i:0 for i in range(num_nodes)}

    for a,b in course:
        graph[b].append(a)
        in_degree[a]+=1

    
    zero_in_degree_nodes=[] # start with an empty dictionary

    for node in range(num_nodes):
        if in_degree[node]==0: # set each nodes in_degree value at 0
            zero_in_degree_nodes.append(node)

    queue=deque(zero_in_degree_nodes)
    order=[]

    while queue:
        node=queue.popleft()
        order.append(node)

        for neighbour in graph[node]:
            in_degree[neighbour]-=1
            if in_degree[neighbour]==0:
                queue.append(neighbour)

    if len(order)!=num_nodes:
        return []

    return order


num_nodes = 4
course = [[1,0], [2,1], [3,2]]
print(course_schedule_kahn_algo(4,graph))
```
