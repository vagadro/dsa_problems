# Flood Fill Algorithm- Solved using DFS

- You're given an image represented as a 2D grid of colors (integers), a starting pixel (sr, sc), and a new color. Fill the starting pixel's color, and every pixel connected to it (up/down/left/right) that shares the same original color, with the new color.

1. Look at the color of the STARTING pixel (sr, sc) -> call this "original_color"

        ↓

2. Safety check: is original_color the SAME as new_color?
        → YES: nothing to do, return the image unchanged (avoids infinite loop)
        → NO: continue

        ↓

3. Start exploring from (sr, sc):

        ↓

   a. Is this position out of bounds? -> STOP, go back
   b. Does this pixel's color NOT match original_color? -> STOP, go back
        (this means we've hit either a different-colored region, or a pixel we already filled)

        ↓

   c. Otherwise: this pixel IS part of our region
        → CHANGE its color to new_color
        → Explore all 4 neighbouring pixels (down, up, right, left) the SAME way

        ↓

4. Keep repeating step 3 outward, in every direction, until every connected
   pixel with the original color has been changed

        ↓

5. Return the fully modified image

```python
image = [
    [1,1,1],
    [1,1,0],
    [1,0,1]
]
```

```python
def flood_fill_dfs(image,sr,sc,new_color):
    rows,cols=len(image), len(image[0])
    original_color=image[sr][sc]

    # If the newcolor is same as original color you stop the recurssion loop otherwise it will get stuck foprever

    if original_color==new_color:
        return image

    def dfs(r,c):
        # Base case: out of bounds
        if r<0 or r>=rows or c<0 or c>=cols or image[r][c]!=original_color:
            return
        
        image[r][c]=new_color #fill the pixel

        dfs(r+1,c)
        dfs(r-1,c)
        dfs(r,c+1)
        dfs(r,c-1)

    dfs(sr,sc)
    return image


result = flood_fill_dfs(image, 1, 1, 2)
for row in result:
    print(row)
```

- Start at (1,1):
  - Is it out of bounds? No.
  - Does it match original_color (1)? Yes.
  - -> FILL it: image[1][1] = 2
  - -> Now explore its 4 neighbours: (2,1), (0,1), (1,2), (1,0)

- Check (2,1): image[2][1] = 0
    - Does NOT match original_color (1) -> STOP here, don't fill, don't explore further

-   Check (0,1): image[0][1] = 1
    - Matches! -> FILL it: image[0][1] = 2
    - -> Explore ITS neighbours: (1,1), (-1,1), (0,2), (0,0)
      -   (1,1): already filled with 2, doesn't match original_color(1) anymore -> STOP
        - (-1,1): out of bounds -> STOP
        - (0,2): matches (1) -> FILL to 2 -> explore further...
        - (0,0): matches (1) -> FILL to 2 -> explore further...

  - Check (1,2): image[1][2] = 0
    - Does NOT match -> STOP

  - Check (1,0): image[1][0] = 1
    - Matches! -> FILL to 2 -> explore its neighbours...
