# Dijkstras Algorithm to calculate shortest path

Find the shortest distance from a source node to every other node, by always expanding outward from whichever unvisited node is currently CLOSEST to the source."

```python
1. Assume every node is infinitely far away (unknown/unreached)
        ↓
2. Except the source itself -> distance 0 (you're already there)
        ↓
3. Put the source into a priority queue (min-heap), tagged with distance 0
        ↓
4. WHILE the priority queue is not empty:
        ↓
   a. Pop the node with the SMALLEST known distance so far
        ↓
   b. Is this distance WORSE than what we already have recorded? 
        → YES: this is a stale/outdated entry, skip it
        → NO: continue
        ↓
   c. For each neighbour of this node:
        - Calculate: "distance to here" + "edge weight to neighbour" = potential new distance
        - Is this potential distance SHORTER than what's currently recorded for the neighbour?
              → YES: update it, and push this neighbour into the queue with its new (better) distance
              → NO: do nothing, we already have a better path
        ↓
5. Once the queue is empty, every node's distance is finalized
        ↓
6. Return the distances dictionary
```

```python
graph = {
    10: [(40,1), (20,1)],
    40: [(10,1),(35,1)],
    35: [(40,1),(50,1),(25,1),(20,1)],
    50: [(35,1)],
    25: [(35,1)],
    20: [(10,1),(35,1),(60,1),(80,1),(70,1)],
    60: [(20,1),(90,1),(80,1),(70,1)],
    80: [(60,1),(20,1),(70,1)],
    70: [(60,1),(20,1),(80,1)],
    90: [(60,1)]
}
```

```python
import heapq
import math

def dijkstras(graph,source):
    distances={node: math.inf for node in graph}
    distances[source]=0
    pq=[(0,source)]

    while pq:
        curr_dis,curr_node=heapq.heappop(pq)

        if curr_dis>distances[curr_node]:
            continue

        for neighbour, weight in graph[curr_node]:
            new_dis=distances[curr_node] + weight
            if new_dis < distances[neighbour]:
                distances[neighbour]=new_dis
                heapq.heappush(pq,(new_dis,neighbour))

    return distances

print(dijkstras(graph, 10))
```
