# Disjoint Union Set (DSU)- Union Find

```python
class DSU:
    def __init__(self,vertices):
        self.parent=list(range(vertices)) #this creates a sequeence parent=[0,1,2,3,4]
        self.rank=[0]*vertices # this basically setsup every node as a tree of height(rank) 0

    # This function is basically finding ouir the root or parent of a particular node by calling recurssive function
    def find(self,node):
        if self.parent[node] != node:
            self.parent[node]=self.find(self.parent[node])
        return self.parent[node]

    #the union function is the one which finds out the hieght(rank) of two nodes or sets and unions it into one. 
    # The set/node with lesser height will be mnerged to the one with larger height or rank

    def union(self,node1,node2):
        root1=self.find(node1)
        root2=self.find(node2)

        #1) If both the nodes belongs to the same parent or root this means roots are already conencted--> union would create a cycle
        if root1==root2:
            return False


        # Union by rank: attach smaler tree under bigger tree
        if self.rank[root1] < self.rank[root2]:
            self.parent[root1]= root2
        elif self.rank[root1] > self.rank[root2]:
            self.parent[root2]=root1
        else:
            self.parent[root2]=root1
            self.rank[root1]+=1

        return True
```

```python
ds = DSU(5)   # nodes 0,1,2,3,4

ds.union(0, 1)
ds.union(2, 3)

print(ds.find(0), ds.find(1))  # same root
print(ds.find(2), ds.find(3))  # same root
print(ds.find(0), ds.find(2))  # different roots

ds.union(1, 2)
print(ds.find(0), ds.find(3))  # now connected
```
