# PRIMS ALGORITHM for Minimum Span Tree (MST)

-  Same goal as Kruskal's: find the cheapest possible set of edges that connects all nodes in a weighted graph, without forming any cycles.

"Start from any one node. Grow the MST one edge at a time, always picking the CHEAPEST edge that connects a node ALREADY in the tree to a node NOT YET in the tree."

```python
import heapq

def prims(graph,start):
    visited=set()
    mst=[]
    total_cost=0

    # Priority queue stores (weight, fromnode, tonode)
    min_heap=[(0, start, start)]

    while min_heap:
        weight, frm, to = heapq.heappop(min_heap)

        if to in visited:
            continue # already part of the MST, skip

        visited.add(to)
        total_cost+= weight

        if frm!=to: #dont append the recode where frmn and to are same
            mst.append((frm,to,weight))

        for neighbour, edge_weight in graph[to]:
            if neighbour not in visited:
                heapq.heappush(min_heap,(edge_weight,to,neighbour))
    return mst,total_cost
```

```python
# Example graph (adjacency list: node -> [(neighbour, weight), ...])
graph = {
    0: [(1, 2), (3, 6)],
    1: [(0, 2), (2, 3), (3, 8), (4, 5)],
    2: [(1, 3), (4, 7)],
    3: [(0, 6), (1, 8), (4, 9)],
    4: [(1, 5), (2, 7), (3, 9)]
}

mst, cost = prims(graph, 0)
print("MST edges:", mst)
print("Total cost:", cost)
```
