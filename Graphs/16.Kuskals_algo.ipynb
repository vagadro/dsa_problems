# KRuskals Algrithm to find MST

```python
1. Take ALL edges in the graph, and SORT them by weight (cheapest first)
        ↓
2. Create a DSU structure — initially, every node is its own separate group
        ↓
3. Create an empty list "mst" to hold the final tree edges, and total_cost = 0
        ↓
4. FOR each edge (weight, u, v), in sorted order (cheapest to most expensive):
        ↓
   a. Ask DSU: "are u and v ALREADY in the same group?"
        ↓
        → YES: Adding this edge would create a CYCLE (redundant connection)
                -> SKIP this edge entirely, move to the next one
        ↓
        → NO: u and v are in DIFFERENT groups
                -> ADD this edge to "mst"
                -> ADD its weight to total_cost
                -> MERGE u's group and v's group together (via dsu.union)
        ↓
   b. Check: have we already collected (V-1) edges?
        → YES: the tree is fully connected — STOP EARLY, no need to check remaining edges
        → NO: continue to the next edge
        ↓
5. Return the final "mst" list and "total_cost"
```
