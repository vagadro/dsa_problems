# Depth First Search

```python
graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 4],
    3: [1],
    4: [1, 2]
}
```

```python
def dfs(graph, source_node, visited=None):
    if visited is None:
        visited=set()

    visited.add(source_node)
    print(source_node, end=" ")

    for neighbour in graph[source_node]:
        if neighbour not in visited:
            dfs(graph,neighbour,visited)

    return visited

dfs(graph,2)
```
