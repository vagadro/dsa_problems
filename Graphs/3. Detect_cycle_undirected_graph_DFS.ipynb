# Detect Cycle in Undirected Graph using DFS



- The logic here is if the currrent-1 node is visited and not parent then there exists a cycle

```python
#Graph 1 is with cycle

graph1 = {
    0: [1, 2],
    1: [0, 2],
    2: [0, 1],
    3: [4],
    4: [3]
}

#Graph 2 is without cycle
graph = {
    0: [1],
    1: [0, 2],
    2: [1]
}
```

```python
def detect_cycle_undirected(graph):
    visited=set()

    def dfs(source_node, parent):
        visited.add(source_node)

        for neighbour in graph[source_node]:
            if neighbour not in visited:
                if dfs(neighbour,source_node):
                    return True
            elif neighbour!=parent:
                return True

        return False

    for node in graph:
        if node not in visited:
            if dfs(node,-1):
                return True
    return False

print(detect_cycle_undirected(graph1))
```
