# DETECT CYCLE IN DIRECTED GRAPH DFS

```python
graph = {
    0: [1],
    1: [2],
    2: [0]   # 2 points back to 0 -> cycle!
}
```

```python
def detect_cycle_directed_dfs(graph):
    visited=set() # nodes filly explored (in any branch, past of present)
    rec_stack=set() # nodes currently in the active recurssion path

    def dfs(node):
        visited.add(node) 
        rec_stack.add(node) #mark as "in progress" on this path

        for neighbour in graph[node]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True
            elif neighbour in rec_stack:
                #found a node that's still on our current path --> cycle deetected!!
                return True

        rec_stack.remove(node) # done exploring tis node's branch, take it OFF the recursionstack
        return False

    for node in graph:
        if node not in visited:
            if dfs(node):
                return True

    return False

print(detect_cycle_directed_dfs(graph))
```
