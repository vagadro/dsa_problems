# Course Scheduling 1- DFS Way- Same as topological order

```python
prerequisites = [[1,0], [2,1], [3,2]]
# Means: 1 needs 0, 2 needs 1, 3 needs 2
```

```python
# In Course Scheduling1 you just find out if you can complete the course- True /False

def can_finish(numCourses,prerequisites):
    graph={i:[] for i in range(numCourses)}
    for a, b in prerequisites:
        graph[b].append(a)


    visited=set()
    rec_stack=set()

    def dfs(node):
        visited.add(node)
        rec_stack.add(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True
            elif neighbour in rec_stack:
                return True

        rec_stack.remove(node)
        return False


    for node in range(numCourses):
        if node not in visited:
            if dfs(node):
                return False

    return True

prerequisites = [[1, 0], [0, 1]]
print(can_finish(2,prerequisites))

prerequisites_ok = [[1, 0], [2, 1], [3, 2]]
print(can_finish(4, prerequisites_ok))  
```
